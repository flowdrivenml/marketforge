```python
import httpx

BASE_URL = "https://api.bybit.com"


def get_tickers(category: str) -> list[dict]:
    response = httpx.get(
        f"{BASE_URL}/v5/market/tickers",
        params={
            "category": category,
        },
        timeout=30.0,
    )

    response.raise_for_status()

    payload = response.json()

    if payload["retCode"] != 0:
        raise RuntimeError(
            f"Bybit error: {payload['retMsg']}"
        )

    return payload["result"]["list"]
```

```python
import httpx


def get_binance_spot_tickers() -> list[dict]:
    response = httpx.get(
        "https://api.binance.com/api/v3/ticker/24hr",
        timeout=30.0,
    )

    response.raise_for_status()

    return response.json()


def get_binance_linear_tickers() -> list[dict]:
    response = httpx.get(
        "https://fapi.binance.com/fapi/v1/ticker/24hr",
        timeout=30.0,
    )

    response.raise_for_status()

    return response.json()


def get_binance_inverse_tickers() -> list[dict]:
    response = httpx.get(
        "https://dapi.binance.com/dapi/v1/ticker/24hr",
        timeout=30.0,
    )

    response.raise_for_status()

    return response.json()
```

```python
spot = get_binance_spot_tickers()
linear = get_binance_linear_tickers()
inverse = get_binance_inverse_tickers()

print("spot:", len(spot))
print("linear:", len(linear))
print("inverse:", len(inverse))
```

```python
from __future__ import annotations

from datetime import datetime, timezone
from decimal import Decimal

import httpx
import pandas as pd

from marketforge.database.connection import connect


# ===========================================================================
# HELPERS
# ===========================================================================

def dec(value) -> Decimal | None:
    if value in (None, ""):
        return None

    return Decimal(str(value))


def zero_dec(value) -> Decimal:
    return dec(value) or Decimal(0)


def get_json(
    url: str,
    *,
    params: dict | None = None,
) -> dict | list:
    response = httpx.get(
        url,
        params=params,
        timeout=30.0,
    )

    response.raise_for_status()

    return response.json()


# One timestamp for this complete refresh.
measured_at = datetime.now(timezone.utc)


# ===========================================================================
# MARKETFORGE INSTRUMENT METADATA
# ===========================================================================

with connect() as conn:
    with conn.cursor() as cursor:
        cursor.execute(
            """
            SELECT
                i.id AS instrument_id,
                e.code AS exchange,
                i.symbol,
                i.instrument_type,
                i.market_category,
                i.base_asset,
                i.quote_asset,
                i.settlement_asset,

                s.quantity_type,
                s.contract_value,
                s.contract_value_asset

            FROM catalog.instruments AS i

            JOIN catalog.exchanges AS e
                ON e.id = i.exchange_id

            LEFT JOIN catalog.instrument_specs AS s
                ON s.instrument_id = i.id

            WHERE i.instrument_type <> 'option'
            """
        )

        instruments = pd.DataFrame(
            cursor.fetchall()
        )


def find_instrument(
    exchange: str,
    symbol: str,
    *,
    instrument_type: str | None = None,
    market_category: str | None = None,
):
    mask = (
        (instruments["exchange"] == exchange)
        & (instruments["symbol"] == symbol)
    )

    if instrument_type is not None:
        mask &= (
            instruments["instrument_type"]
            == instrument_type
        )

    if market_category is not None:
        mask &= (
            instruments["market_category"]
            == market_category
        )

    rows = instruments[mask]

    if rows.empty:
        return None

    if len(rows) > 1:
        raise RuntimeError(
            "Ambiguous instrument lookup: "
            f"{exchange=} {symbol=} "
            f"{instrument_type=} "
            f"{market_category=}"
        )

    return rows.iloc[0]


# ===========================================================================
# CANONICAL MARKET SNAPSHOT
# ===========================================================================

market_rows: list[dict] = []


def add_market_snapshot(
    *,
    instrument,
    last_price: Decimal | None = None,
    high_24h: Decimal | None = None,
    low_24h: Decimal | None = None,
    price_change_pct_24h: Decimal | None = None,

    volume_24h_native: Decimal | None = None,
    volume_24h_base: Decimal | None = None,
    volume_24h_quote: Decimal | None = None,
    volume_24h_contracts: Decimal | None = None,

    turnover_24h: Decimal | None = None,
    turnover_denomination: str | None = None,

    open_interest: Decimal | None = None,
    open_interest_value: Decimal | None = None,
    funding_rate: Decimal | None = None,

    source: str,
) -> None:
    market_rows.append(
        {
            "instrument_id": int(
                instrument["instrument_id"]
            ),

            "exchange": instrument["exchange"],
            "symbol": instrument["symbol"],

            "instrument_type": (
                instrument["instrument_type"]
            ),

            "market_category": (
                instrument["market_category"]
            ),

            "base_asset": (
                instrument["base_asset"]
            ),

            "quote_asset": (
                instrument["quote_asset"]
            ),

            "settlement_asset": (
                instrument["settlement_asset"]
            ),

            "last_price": last_price,
            "high_24h": high_24h,
            "low_24h": low_24h,

            "price_change_pct_24h": (
                price_change_pct_24h
            ),

            "volume_24h_native": (
                volume_24h_native
            ),

            "volume_24h_base": (
                volume_24h_base
            ),

            "volume_24h_quote": (
                volume_24h_quote
            ),

            "volume_24h_contracts": (
                volume_24h_contracts
            ),

            "turnover_24h": turnover_24h,

            "turnover_denomination": (
                turnover_denomination
            ),

            "open_interest": open_interest,

            "open_interest_value": (
                open_interest_value
            ),

            "funding_rate": funding_rate,

            "measured_at": measured_at,
            "source": source,
        }
    )


# ===========================================================================
# BYBIT
# ===========================================================================

def get_bybit_tickers(
    category: str,
) -> list[dict]:
    payload = get_json(
        "https://api.bybit.com/v5/market/tickers",
        params={
            "category": category,
        },
    )

    if payload["retCode"] != 0:
        raise RuntimeError(
            f"Bybit: {payload['retMsg']}"
        )

    return payload["result"]["list"]


bybit_spot = get_bybit_tickers("spot")
bybit_linear = get_bybit_tickers("linear")
bybit_inverse = get_bybit_tickers("inverse")


# Spot:
# volume24h   = base quantity
# turnover24h = quote turnover

for row in bybit_spot:
    instrument = find_instrument(
        "bybit",
        row["symbol"],
        instrument_type="spot",
        market_category="spot",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice24h")
        ),

        low_24h=dec(
            row.get("lowPrice24h")
        ),

        price_change_pct_24h=dec(
            row.get("price24hPcnt")
        ),

        volume_24h_native=dec(
            row.get("volume24h")
        ),

        volume_24h_base=dec(
            row.get("volume24h")
        ),

        volume_24h_quote=dec(
            row.get("turnover24h")
        ),

        turnover_24h=dec(
            row.get("turnover24h")
        ),

        turnover_denomination=(
            instrument["quote_asset"]
        ),

        source="bybit:spot",
    )


# Linear:
# volume24h   = base quantity
# turnover24h = quote/USD-like turnover

for row in bybit_linear:
    instrument = find_instrument(
        "bybit",
        row["symbol"],
        market_category="linear",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice24h")
        ),

        low_24h=dec(
            row.get("lowPrice24h")
        ),

        price_change_pct_24h=dec(
            row.get("price24hPcnt")
        ),

        volume_24h_native=dec(
            row.get("volume24h")
        ),

        volume_24h_base=dec(
            row.get("volume24h")
        ),

        volume_24h_quote=dec(
            row.get("turnover24h")
        ),

        turnover_24h=dec(
            row.get("turnover24h")
        ),

        turnover_denomination="USD",

        open_interest=dec(
            row.get("openInterest")
        ),

        open_interest_value=dec(
            row.get("openInterestValue")
        ),

        funding_rate=dec(
            row.get("fundingRate")
        ),

        source="bybit:linear",
    )


# Inverse:
# volume24h   = USD-like contract notional
# turnover24h = base turnover

for row in bybit_inverse:
    instrument = find_instrument(
        "bybit",
        row["symbol"],
        market_category="inverse",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice24h")
        ),

        low_24h=dec(
            row.get("lowPrice24h")
        ),

        price_change_pct_24h=dec(
            row.get("price24hPcnt")
        ),

        volume_24h_native=dec(
            row.get("volume24h")
        ),

        volume_24h_base=dec(
            row.get("turnover24h")
        ),

        volume_24h_quote=dec(
            row.get("volume24h")
        ),

        turnover_24h=dec(
            row.get("volume24h")
        ),

        turnover_denomination="USD",

        open_interest=dec(
            row.get("openInterest")
        ),

        open_interest_value=dec(
            row.get("openInterestValue")
        ),

        funding_rate=dec(
            row.get("fundingRate")
        ),

        source="bybit:inverse",
    )


# ===========================================================================
# BINANCE
# ===========================================================================

binance_spot = get_json(
    "https://api.binance.com/api/v3/ticker/24hr"
)

binance_linear = get_json(
    "https://fapi.binance.com/fapi/v1/ticker/24hr"
)

binance_inverse = get_json(
    "https://dapi.binance.com/dapi/v1/ticker/24hr"
)


# Spot

for row in binance_spot:
    instrument = find_instrument(
        "binance",
        row["symbol"],
        instrument_type="spot",
        market_category="spot",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice")
        ),

        low_24h=dec(
            row.get("lowPrice")
        ),

        price_change_pct_24h=(
            dec(row.get("priceChangePercent"))
            / Decimal(100)
            if dec(row.get("priceChangePercent"))
            is not None
            else None
        ),

        volume_24h_native=dec(
            row.get("volume")
        ),

        volume_24h_base=dec(
            row.get("volume")
        ),

        volume_24h_quote=dec(
            row.get("quoteVolume")
        ),

        turnover_24h=dec(
            row.get("quoteVolume")
        ),

        turnover_denomination=(
            instrument["quote_asset"]
        ),

        source="binance:spot",
    )


# Linear

for row in binance_linear:
    instrument = find_instrument(
        "binance",
        row["symbol"],
        market_category="linear",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice")
        ),

        low_24h=dec(
            row.get("lowPrice")
        ),

        price_change_pct_24h=(
            dec(row.get("priceChangePercent"))
            / Decimal(100)
            if dec(row.get("priceChangePercent"))
            is not None
            else None
        ),

        volume_24h_native=dec(
            row.get("volume")
        ),

        volume_24h_base=dec(
            row.get("volume")
        ),

        volume_24h_quote=dec(
            row.get("quoteVolume")
        ),

        turnover_24h=dec(
            row.get("quoteVolume")
        ),

        turnover_denomination="USD",

        source="binance:linear",
    )


# Inverse

for row in binance_inverse:
    instrument = find_instrument(
        "binance",
        row["symbol"],
        market_category="inverse",
    )

    if instrument is None:
        continue

    contract_value = (
        instrument["contract_value"]
    )

    contract_value_asset = (
        instrument["contract_value_asset"]
    )

    volume_contracts = dec(
        row.get("volume")
    )

    base_volume = dec(
        row.get("baseVolume")
    )

    turnover = None

    if (
        contract_value is not None
        and not pd.isna(contract_value)
        and contract_value_asset == "USD"
        and volume_contracts is not None
    ):
        turnover = (
            volume_contracts
            * dec(contract_value)
        )

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice")
        ),

        low_24h=dec(
            row.get("lowPrice")
        ),

        price_change_pct_24h=(
            dec(row.get("priceChangePercent"))
            / Decimal(100)
            if dec(row.get("priceChangePercent"))
            is not None
            else None
        ),

        volume_24h_native=volume_contracts,

        volume_24h_base=base_volume,

        volume_24h_quote=turnover,

        volume_24h_contracts=volume_contracts,

        turnover_24h=turnover,

        turnover_denomination="USD",

        source="binance:inverse",
    )


# ===========================================================================
# OKX
# ===========================================================================

def get_okx_tickers(
    inst_type: str,
) -> list[dict]:
    payload = get_json(
        "https://www.okx.com/api/v5/market/tickers",
        params={
            "instType": inst_type,
        },
    )

    if payload["code"] != "0":
        raise RuntimeError(
            f"OKX: {payload['msg']}"
        )

    return payload["data"]


okx_spot = get_okx_tickers("SPOT")
okx_swap = get_okx_tickers("SWAP")
okx_futures = get_okx_tickers("FUTURES")


# Spot

for row in okx_spot:
    instrument = find_instrument(
        "okx",
        row["instId"],
        instrument_type="spot",
        market_category="spot",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("last")
        ),

        high_24h=dec(
            row.get("high24h")
        ),

        low_24h=dec(
            row.get("low24h")
        ),

        volume_24h_native=dec(
            row.get("vol24h")
        ),

        volume_24h_base=dec(
            row.get("vol24h")
        ),

        volume_24h_quote=dec(
            row.get("volCcy24h")
        ),

        turnover_24h=dec(
            row.get("volCcy24h")
        ),

        turnover_denomination=(
            instrument["quote_asset"]
        ),

        source="okx:spot",
    )


# SWAP + FUTURES

for row in okx_swap + okx_futures:
    instrument = find_instrument(
        "okx",
        row["instId"],
    )

    if instrument is None:
        continue

    contracts = dec(
        row.get("vol24h")
    )

    volume_ccy = dec(
        row.get("volCcy24h")
    )

    last_price = dec(
        row.get("last")
    )

    contract_value = (
        instrument["contract_value"]
    )

    contract_value_asset = (
        instrument["contract_value_asset"]
    )

    turnover = None
    base_volume = None

    if (
        contracts is not None
        and contract_value is not None
        and not pd.isna(contract_value)
    ):
        contract_value_dec = dec(
            contract_value
        )

        if contract_value_asset == "USD":
            # Inverse contract.
            turnover = (
                contracts
                * contract_value_dec
            )

            base_volume = volume_ccy

        elif (
            contract_value_asset
            == instrument["base_asset"]
        ):
            # Linear contract.
            base_volume = (
                contracts
                * contract_value_dec
            )

            if last_price is not None:
                turnover = (
                    base_volume
                    * last_price
                )

    add_market_snapshot(
        instrument=instrument,

        last_price=last_price,

        high_24h=dec(
            row.get("high24h")
        ),

        low_24h=dec(
            row.get("low24h")
        ),

        volume_24h_native=contracts,

        volume_24h_base=base_volume,

        volume_24h_quote=turnover,

        volume_24h_contracts=contracts,

        turnover_24h=turnover,

        turnover_denomination="USD",

        source=(
            f"okx:{row['instType'].lower()}"
        ),
    )


# ===========================================================================
# BITGET
# ===========================================================================

def get_bitget_tickers(
    category: str,
) -> list[dict]:
    payload = get_json(
        "https://api.bitget.com/api/v3/market/tickers",
        params={
            "category": category,
        },
    )

    if payload["code"] != "00000":
        raise RuntimeError(
            f"Bitget: {payload['msg']}"
        )

    return payload["data"]


bitget_spot = get_bitget_tickers(
    "SPOT"
)

bitget_usdt = get_bitget_tickers(
    "USDT-FUTURES"
)

bitget_usdc = get_bitget_tickers(
    "USDC-FUTURES"
)

bitget_coin = get_bitget_tickers(
    "COIN-FUTURES"
)


# Spot

for row in bitget_spot:
    instrument = find_instrument(
        "bitget",
        row["symbol"],
        instrument_type="spot",
        market_category="spot",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice24h")
        ),

        low_24h=dec(
            row.get("lowPrice24h")
        ),

        price_change_pct_24h=dec(
            row.get("price24hPcnt")
        ),

        volume_24h_native=dec(
            row.get("volume24h")
        ),

        volume_24h_base=dec(
            row.get("volume24h")
        ),

        volume_24h_quote=dec(
            row.get("turnover24h")
        ),

        turnover_24h=dec(
            row.get("turnover24h")
        ),

        turnover_denomination=(
            instrument["quote_asset"]
        ),

        source="bitget:spot",
    )


# Linear USDT + USDC

for rows, source in (
    (
        bitget_usdt,
        "bitget:usdt-futures",
    ),
    (
        bitget_usdc,
        "bitget:usdc-futures",
    ),
):
    for row in rows:
        instrument = find_instrument(
            "bitget",
            row["symbol"],
            market_category="linear",
        )

        if instrument is None:
            continue

        add_market_snapshot(
            instrument=instrument,

            last_price=dec(
                row.get("lastPrice")
            ),

            high_24h=dec(
                row.get("highPrice24h")
            ),

            low_24h=dec(
                row.get("lowPrice24h")
            ),

            price_change_pct_24h=dec(
                row.get("price24hPcnt")
            ),

            volume_24h_native=dec(
                row.get("volume24h")
            ),

            volume_24h_base=dec(
                row.get("volume24h")
            ),

            volume_24h_quote=dec(
                row.get("turnover24h")
            ),

            turnover_24h=dec(
                row.get("turnover24h")
            ),

            turnover_denomination="USD",

            open_interest=dec(
                row.get("openInterest")
            ),

            funding_rate=dec(
                row.get("fundingRate")
            ),

            source=source,
        )


# COIN / inverse

for row in bitget_coin:
    instrument = find_instrument(
        "bitget",
        row["symbol"],
        market_category="inverse",
    )

    if instrument is None:
        continue

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("lastPrice")
        ),

        high_24h=dec(
            row.get("highPrice24h")
        ),

        low_24h=dec(
            row.get("lowPrice24h")
        ),

        price_change_pct_24h=dec(
            row.get("price24hPcnt")
        ),

        volume_24h_native=dec(
            row.get("volume24h")
        ),

        volume_24h_base=dec(
            row.get("volume24h")
        ),

        volume_24h_quote=dec(
            row.get("turnover24h")
        ),

        turnover_24h=dec(
            row.get("turnover24h")
        ),

        turnover_denomination="USD",

        open_interest=dec(
            row.get("openInterest")
        ),

        funding_rate=dec(
            row.get("fundingRate")
        ),

        source="bitget:coin-futures",
    )


# ===========================================================================
# GATE.IO
# ===========================================================================

GATE_BASE_URL = "https://api.gateio.ws/api/v4"


gate_spot = get_json(
    f"{GATE_BASE_URL}/spot/tickers"
)


def get_gate_futures(
    settle: str,
) -> list[dict]:
    return get_json(
        f"{GATE_BASE_URL}/futures/{settle}/tickers"
    )


gate_usdt = get_gate_futures("usdt")
gate_btc = get_gate_futures("btc")
gate_usd1 = get_gate_futures("usd1")


# Spot

for row in gate_spot:
    instrument = find_instrument(
        "gateio",
        row["currency_pair"],
        instrument_type="spot",
        market_category="spot",
    )

    if instrument is None:
        continue

    change_pct = dec(
        row.get("change_percentage")
    )

    if change_pct is not None:
        change_pct /= Decimal(100)

    add_market_snapshot(
        instrument=instrument,

        last_price=dec(
            row.get("last")
        ),

        high_24h=dec(
            row.get("high_24h")
        ),

        low_24h=dec(
            row.get("low_24h")
        ),

        price_change_pct_24h=change_pct,

        volume_24h_native=dec(
            row.get("base_volume")
        ),

        volume_24h_base=dec(
            row.get("base_volume")
        ),

        volume_24h_quote=dec(
            row.get("quote_volume")
        ),

        turnover_24h=dec(
            row.get("quote_volume")
        ),

        turnover_denomination=(
            instrument["quote_asset"]
        ),

        source="gateio:spot",
    )


# Contracts

# Contracts

for rows, source, market_category in (
    (
        gate_usdt,
        "gateio:usdt-futures",
        "linear",
    ),
    (
        gate_btc,
        "gateio:btc-futures",
        "inverse",
    ),
    (
        gate_usd1,
        "gateio:usd1-futures",
        "linear",
    ),
):
    for row in rows:
        instrument = find_instrument(
            "gateio",
            row["contract"],
            market_category=market_category,
        )

        if instrument is None:
            continue

        change_pct = dec(
            row.get("change_percentage")
        )

        if change_pct is not None:
            change_pct /= Decimal(100)

        add_market_snapshot(
            instrument=instrument,

            last_price=dec(
                row.get("last")
            ),

            high_24h=dec(
                row.get("high_24h")
            ),

            low_24h=dec(
                row.get("low_24h")
            ),

            price_change_pct_24h=change_pct,

            volume_24h_native=dec(
                row.get("volume_24h")
            ),

            volume_24h_base=dec(
                row.get("volume_24h_base")
            ),

            volume_24h_quote=dec(
                row.get("volume_24h_quote")
            ),

            volume_24h_contracts=dec(
                row.get("volume_24h")
            ),

            turnover_24h=dec(
                row.get("volume_24h_quote")
            ),

            turnover_denomination="USD",

            funding_rate=dec(
                row.get("funding_rate")
            ),

            source=source,
        )


# ===========================================================================
# BUILD MARKET METRICS TABLE
# ===========================================================================

metrics = pd.DataFrame(
    market_rows
)


# ===========================================================================
# DERIVED RANKINGS
# ===========================================================================

# Contracts can be ranked together because turnover_24h has been
# normalized to approximate USD-equivalent notional.

contracts = metrics[
    metrics["market_category"].isin(
        [
            "linear",
            "inverse",
        ]
    )
].copy()

contracts = contracts[
    contracts["turnover_24h"].notna()
].copy()

contracts = contracts.sort_values(
    [
        "turnover_24h",
        "exchange",
        "symbol",
    ],
    ascending=[
        False,
        True,
        True,
    ],
).reset_index(drop=True)

contracts["turnover_rank"] = (
    contracts.index + 1
)


# Spot turnover is only directly comparable inside the same
# quote denomination.

spot = metrics[
    metrics["market_category"] == "spot"
].copy()

spot = spot[
    spot["turnover_24h"].notna()
].copy()

spot = spot.sort_values(
    [
        "turnover_denomination",
        "turnover_24h",
        "exchange",
        "symbol",
    ],
    ascending=[
        True,
        False,
        True,
        True,
    ],
).reset_index(drop=True)

spot["turnover_rank"] = (
    spot.groupby(
        "turnover_denomination",
        sort=False,
    )
    .cumcount()
    + 1
)


# ===========================================================================
# COVERAGE
# ===========================================================================

print(
    f"Catalog instruments : "
    f"{len(instruments):,}"
)

print(
    f"Metric snapshots    : "
    f"{len(metrics):,}"
)

print(
    f"Contract snapshots  : "
    f"{len(contracts):,}"
)

print(
    f"Spot snapshots      : "
    f"{len(spot):,}"
)

print()

print("Coverage by exchange / market:")
display(
    metrics.groupby(
        [
            "exchange",
            "instrument_type",
            "market_category",
        ],
        dropna=False,
    )
    .size()
    .rename("count")
    .reset_index()
)


# ===========================================================================
# TOP CONTRACTS BY TURNOVER
# ===========================================================================

print("Top 50 contracts by normalized 24h turnover:")

display(
    contracts[
        [
            "turnover_rank",
            "instrument_id",
            "exchange",
            "symbol",
            "instrument_type",
            "market_category",
            "last_price",
            "turnover_24h",
            "turnover_denomination",
            "volume_24h_base",
            "volume_24h_contracts",
            "open_interest",
            "open_interest_value",
            "funding_rate",
            "source",
        ]
    ].head(50)
)


# ===========================================================================
# TOP USDT SPOT MARKETS
# ===========================================================================

print("Top 50 USDT spot markets by 24h turnover:")

display(
    spot[
        spot["turnover_denomination"]
        == "USDT"
    ][
        [
            "turnover_rank",
            "instrument_id",
            "exchange",
            "symbol",
            "base_asset",
            "quote_asset",
            "last_price",
            "turnover_24h",
            "volume_24h_base",
            "high_24h",
            "low_24h",
            "price_change_pct_24h",
            "source",
        ]
    ].head(50)
)


# ===========================================================================
# TOP MARKETS PER EXCHANGE
# ===========================================================================

print("Top 10 contracts per exchange:")

top_contracts_per_exchange = (
    contracts
    .sort_values(
        "turnover_24h",
        ascending=False,
    )
    .groupby(
        "exchange",
        group_keys=False,
    )
    .head(10)
)

display(
    top_contracts_per_exchange[
        [
            "exchange",
            "symbol",
            "market_category",
            "turnover_24h",
            "turnover_denomination",
            "volume_24h_base",
            "open_interest",
            "funding_rate",
        ]
    ]
)


# ===========================================================================
# EXCHANGE SUMMARY
# ===========================================================================

print("Contract activity by exchange:")

exchange_summary = (
    contracts
    .groupby(
        "exchange"
    )
    .agg(
        instruments=(
            "instrument_id",
            "count",
        ),
        total_turnover_24h=(
            "turnover_24h",
            "sum",
        ),
        median_turnover_24h=(
            "turnover_24h",
            "median",
        ),
        max_turnover_24h=(
            "turnover_24h",
            "max",
        ),
    )
    .sort_values(
        "total_turnover_24h",
        ascending=False,
    )
    .reset_index()
)

display(exchange_summary)


# ===========================================================================
# RAW METRICS TABLE
# ===========================================================================

print("Full canonical market snapshot:")

display(
    metrics.sort_values(
        [
            "exchange",
            "market_category",
            "symbol",
        ]
    )
)
```
