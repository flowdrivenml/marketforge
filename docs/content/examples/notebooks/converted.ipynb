# Inspect converted market values

Use an existing converted combined publication. The Bybit Quick Start stops after merge. Open this notebook inside your checkout with the MarketForge research kernel. It reads existing files and never runs conversion.

```python
from forge.datasets import MarketData

market = MarketData()
catalog = market.discover(metadata_only=True)

quote = input("Target quote (for example, USDT): ").strip().upper()
matches = catalog.find(operation="convert", target_quote=quote)
display(matches.summary())
```

Choose the publication you want to inspect. In Forge, inspect **Data → Converted** and use the directory containing the recorded manifest. The indexed output reference identifies a registry record, while this notebook opens the filesystem path.

```python
import json
from pathlib import Path

output = market.open(input("Converted publication directory: ").strip())
display(output.info())
display(output.schema())
display(output.lineage())
manifest_path = Path(output.info()["publication_path"]) / "manifest.json"
manifest = json.loads(manifest_path.read_text(encoding="utf-8"))
display(manifest["conversion"])
```

Review the target quote and conversion policy before interpreting target-valued fields. Exact decimal values appear as strings in a preview. Arrow batches retain their types.

```python
trades = output.query(data_type="trades")
display(trades.head(5, input_rows=20_000))
for batch in trades.batches(batch_size=1024, input_rows=20_000):
    print(batch.schema)
    print(batch.num_rows)
    break
```
