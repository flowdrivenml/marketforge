# Inspect normalized market data

Use the MarketForge research kernel and run this notebook inside the checkout. You need an existing normalized publication. The Bybit Quick Start produces normalized files before merge. These cells read local files and require no registry connection.

```python
from forge.datasets import MarketData

market = MarketData()
catalog = market.discover(metadata_only=True)

normalized = catalog.find(operation="normalize", exchange="bybit", symbol="BTCUSDT")
display(normalized.find(data_type="trades").summary())
display(normalized.find(data_type="depth").summary())
```

Choose an exact file reference from either table. Change the market filters if you have data from another instrument. If the tables are empty, check your workspace and storage root.

```python
dataset = normalized.get(input("Exact normalized file reference: ").strip())
display(dataset.discovery_fields())
display(dataset.info())
display(dataset.schema())
```

Enter an inclusive start and exclusive end in UTC for the selected file. The preview reads only a bounded physical prefix. A range late in the file can produce an empty result.

```python
start = input("Start UTC (ISO timestamp): ").strip()
end = input("End UTC (ISO timestamp): ").strip()
display(dataset.head(
    5, columns=["event_timestamp_ns", "symbol"],
    start=start, end=end, input_rows=100_000,
))
```
