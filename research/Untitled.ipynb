```python
from forge.datasets import MarketData
from forge.lab.inspection import inspect

mf = MarketData()
```

```python
catalog = mf.discover(scope="tests", include_staging=True)
```

```python
dataset = mf.open(
    "data/tests/historical/normalize/full-run/"
    "20261115-202550/staging/"
    "dataset-1335781940318399008",
    allow_staging=True,
)

display(dataset.streams())
```

```python
stream = dataset.select(
    fingerprint=(
        "80e5604cd6e92cc519932512e3f9bdd"
        "fafc353aae20352e676aca2bd0de99c1c"
    )
)

display(stream.head(20))
```

```python
report = inspect(stream)

print(report.schema())
print(report.summary())
```

```python
import polars as pl

lf = stream.scan(
    columns=[
        "event_timestamp_ns",
        "exchange",
        "symbol",
        "trade_id",
        "side",
    ]
)

display(
    lf.group_by("side")
    .agg(pl.len().alias("trades"))
    .collect()
)
```
