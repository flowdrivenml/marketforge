# MarketForge — Databento NASDAQ Equities Sample

Download **AAPL trade ticks** and **10-level NASDAQ order-book (MBP-10) updates**, keeping both the compact original DBN and an inspectable JSON Lines version.

**Requirements:** A Databento account/API key and historical access to `XNAS.ITCH`. Usage may consume trial credits or incur charges. The notebook estimates the cost and does **not** download until you explicitly enable it.

**Important:** `mbp-10` is Databento's normalized *market-by-price event feed with ten visible price levels*, not the same snapshot/delta protocol as a crypto venue. A request starting at 14:30 UTC does not necessarily contain all prior updates required to reconstruct the book from scratch; interpret the provided levels and event fields accordingly. Only NASDAQ venue data is included.

## 2. Authentication (key stays out of notebook source)

```python
from getpass import getpass
import databento as db

api_key = getpass("Databento API key: ")
client = db.Historical(api_key)
print("Historical client initialized.")
```

## 3. Request parameters

```python
from pathlib import Path

DATASET = "XNAS.ITCH"
SYMBOL = "AAPL"
START = "2024-10-28T14:30:00Z"  # 09:30 New York, EDT
END = "2024-10-28T14:40:00Z"
SCHEMAS = {"trades": "trades", "mbp-10": "orderbook_L2"}
OUTPUT = Path("databento_samples") / SYMBOL / "2024-10-28"
OUTPUT.mkdir(parents=True, exist_ok=True)
print(f"Dataset: {DATASET} | Symbol: {SYMBOL} | UTC: {START} to {END}")
```

## 4. Check schema availability and estimated charges (no download)

```python
available = set(client.metadata.list_schemas(dataset=DATASET))
missing = set(SCHEMAS) - available
if missing:
    raise RuntimeError(f"Unsupported schemas: {sorted(missing)}. Available: {sorted(available)}")

costs = {}
for schema in SCHEMAS:
    costs[schema] = client.metadata.get_cost(
        dataset=DATASET,
        symbols=SYMBOL,
        schema=schema,
        start=START,
        end=END,
    )
    print(f"{schema:>8}: estimated ${costs[schema]:.4f}")

print(f"TOTAL estimated: ${sum(costs.values()):.4f}")
print("Estimates may differ from actual billing. Check your account's credits/entitlements.")
```

## 5. Download (explicit opt-in)

Change `DOWNLOAD = False` to `True` **only after reviewing the estimates**. Data is streamed to files rather than loaded into notebook memory.

```python
DOWNLOAD = False  # change to True to approve the estimated cost

if not DOWNLOAD:
    print("No download performed. Set DOWNLOAD=True after reviewing the cost estimate.")
else:
    for schema, label in SCHEMAS.items():
        dbn_path = OUTPUT / f"{SYMBOL}_{label}.dbn.zst"
        jsonl_path = OUTPUT / f"{SYMBOL}_{label}.jsonl"
        print(f"Fetching {schema} -> {dbn_path}")
        store = client.timeseries.get_range(
            dataset=DATASET,
            symbols=SYMBOL,
            schema=schema,
            start=START,
            end=END,
            path=dbn_path,
        )
        store.to_json(jsonl_path)
        print(f"Saved {jsonl_path} ({jsonl_path.stat().st_size:,} bytes)")
```

## 6. Inspect the first trade and order-book events

```python
import json
from IPython.display import display, JSON

for label in SCHEMAS.values():
    path = OUTPUT / f"{SYMBOL}_{label}.jsonl"
    print(f"\n{path}")
    if not path.exists():
        print("Not downloaded yet.")
        continue
    with path.open(encoding="utf-8") as f:
        for _ in range(2):
            line = f.readline()
            if not line:
                break
            display(JSON(json.loads(line), expanded=False))
```

## 7. Preview records as tables without reading the entire file

```python
import pandas as pd

for label in SCHEMAS.values():
    path = OUTPUT / f"{SYMBOL}_{label}.jsonl"
    if path.exists():
        print(f"\n{label}")
        display(pd.read_json(path, lines=True, nrows=5).T)
```

## Notes for MarketForge

- **Trades** represent NASDAQ executions in this feed, not consolidated trades from all US stock exchanges.
- **MBP-10** gives 10 bid and ask levels with event context; do not assume its records are complete independent snapshots unless the documentation/flags and state initialization support that interpretation.
- `DBN` is the efficient archival format. JSONL is convenient for inspection, but can be substantially larger.
- Keep original records if you plan to derive MarketForge-specific trade/depth schemas; retain event sequence, publisher, instrument and timestamps.

Official references: [XNAS.ITCH dataset](https://databento.com/datasets/XNAS.ITCH), [Python historical API](https://databento.com/docs/api-historical?lang=python), [get_cost](https://databento.com/docs/api-historical/metadata/metadata-get-cost?lang=python).
