```python
from forge.datasets import MarketData

mf = MarketData()
catalog = mf.discover(scope="committed")

normalized = catalog.find(operation="normalize")

normalized.overview()
```

```python
normalized.summary()
```

```python
trades = normalized.find(data_type="trades")
depth = normalized.find(data_type="depth")

trades.summary()
```

```python
selected = normalized.find(
    exchange="bybit",
    symbol="BTCUSDT",
    instrument_type="perpetual",
    data_type="trades",
)

selected.summary()
```

```python
dataset = selected[0]

dataset.info()
```
